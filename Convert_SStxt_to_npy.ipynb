```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Read Input
def read_variables_to_globals(filename):
    with open(filename, 'r') as f:
        lines = [line.strip() for line in f.readlines()]

    for i in range(0, len(lines), 2):
        var_name = lines[i]
        raw_value = lines[i+1]

        try:
            value = int(raw_value)
        except ValueError:
            try:
                value = float(raw_value)
            except ValueError:
                value = raw_value

        globals()[var_name] = value  # Define variable in global scope
```

```python
def convert_to_npy(ProjectName, NumSS):
    # -------------------------------------------------------------------
    # Loop through all snapshots and convert .txt → .npy
    # -------------------------------------------------------------------
    for i in range(NumSS):
        snap_id = i + 1
        snap_filename_txt = f"{ProjectName}_SnapShot_{snap_id}.txt"
        snap_filename_npy = f"{ProjectName}_SnapShot_{snap_id}.npy"

        if not os.path.exists(snap_filename_txt):
            print(f"  Skipping missing file: {snap_filename_txt}")
            continue

        print(f"Reading {snap_filename_txt} ...")
        try:
            data = np.loadtxt(snap_filename_txt)
        except Exception as e:
            print(f"Error reading {snap_filename_txt}: {e}")
            continue

        print(f"Saving {snap_filename_npy} ...")
        np.save(snap_filename_npy, data)

    print("\n Conversion complete: all available .txt snapshots saved as .npy")
```
