```python
# This code basically takes in a group of snapshots and then subtracts the average
```

```python
import numpy as np
```

```python

def subtract_snapshot_average(ProjectName, NumSS):
    """
    Reads snapshots from text files, computes their average,
    subtracts the average from each snapshot, and saves the results.
    
    Parameters:
    - ProjectName (str): Base name of snapshot files (e.g., "ToyData")
    - NumSS (int): Number of snapshots
    """
    
    # ----- First loop: compute average snapshot -----
    sum_snapshot = None

    for i in range(1, NumSS + 1):
        snap_filename = f"{ProjectName}_SnapShot_{i}.txt"
        snapshot = np.loadtxt(snap_filename)
        
        if sum_snapshot is None:
            sum_snapshot = np.zeros_like(snapshot, dtype=float)
        
        sum_snapshot += snapshot
    
    avg_snapshot = sum_snapshot / NumSS
    save_avg_filename = f"{ProjectName}_Avg_Flowfield.txt"
    np.savetxt(save_avg_filename, avg_snapshot, fmt="%.6e")
    print(f"Saved averaged flowfield to {save_avg_filename}")

    # ----- Second loop: subtract average and save -----
    for i in range(1, NumSS + 1):
        snap_filename = f"{ProjectName}_SnapShot_{i}.txt"
        snapshot = np.loadtxt(snap_filename)
        snapshot_centered = snapshot - avg_snapshot
        
        output_filename = f"{ProjectName}_Avg_SnapShot_{i}.txt"
        np.savetxt(output_filename, snapshot_centered, fmt="%.6e")
        print(f"Saved centered snapshot to {output_filename}")
```

```python
if __name__ == "__main__":

    subtract_snapshot_average("Vortex", 151)
```
